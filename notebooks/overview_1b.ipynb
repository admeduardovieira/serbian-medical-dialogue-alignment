# 1B IT: response analysis

Baseline, SFT, and DPO responses on 315 validation turns. The baseline, SFT, and DPO models all start from the 1B IT checkpoint.

Each example includes the conversation context, recorded reference response, BERTScore F1, and the four judge ratings. Conversations, responses, and judge explanations remain in their original language.

Aggregate tables and plots are in the [model evaluation](results.ipynb) notebook.

```python
from IPython.display import display

import story_utils as S

items = S.load_items()

MODELS = S.SIZE_GROUPS["1b"]
```

## Selected examples

These examples were selected manually to inspect differences between SFT and DPO, including disagreements between BERTScore and judge ratings and a case where the judge rates DPO lower. They illustrate individual behaviors, not their frequency in the validation set.

```python
HIGHLIGHTS = [
    ('008-record-1731398023467_ms_gm_ms', 9,
     'The SFT response asks whether the conversation concerns a child, which the context already establishes. DPO asks about symptom history. The judge rates DPO higher on all four rubrics, although SFT has the higher BERTScore.'),
    ('024_TamaraTijanaKašalj_ms', 15,
     'SFT suggests an intervention instead of collecting information. DPO asks a follow-up question and receives a higher safety and role-adherence rating.'),
    ('058_record-1734430536727_ms', 11,
     'The parent has already described the rash as red. SFT asks about its color again; DPO asks whether this has happened before.'),
    ('057_record_241217_ms', 19,
     'SFT returns to vaccination history that has already been discussed. DPO asks about recent medication and receives a higher clinical-usefulness rating.'),
    ('115_record-1738939856383_ms', 1,
     'The brief SFT greeting closely matches the reference and has a higher BERTScore. DPO adds a service introduction and receives a higher mean judge score. The two measures favor different responses.'),
    ('130_record-1740145201901_ms', 19,
     'Here DPO repeats a question about a head injury that the conversation has already addressed. SFT asks about the nose. This example shows a decrease in the judge ratings after DPO.'),
]

for call_id, turn_id, caption in HIGHLIGHTS:
    display(S.show_item(
        items, call_id, turn_id, models=MODELS[1:],
        metrics=["bertscore_f1"], caption=caption,
    ))
```

Examples are ordered by conversation and turn. The full output is intentionally kept in this notebook so the aggregate report remains short.

```python
ordered_items = sorted(items, key=lambda item: (item["call_id"], item["turn_id"]))

for item in ordered_items:
    display(S.show_item(
        items, item["call_id"], item["turn_id"],
        models=MODELS, metrics=["bertscore_f1"],
    ))
```
