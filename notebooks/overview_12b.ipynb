# 12B PT: response analysis

Baseline, SFT, and DPO responses on 315 validation turns. The SFT and DPO adapters start from 12B PT. The baseline uses 12B IT; it is not the unadapted PT checkpoint.

Each example includes the conversation context, recorded reference response, BERTScore F1, and the four judge ratings. Conversations, responses, and judge explanations remain in their original language.

Aggregate tables and plots are in the [model evaluation](results.ipynb) notebook.

```python
from IPython.display import display

import story_utils as S

items = S.load_items()

MODELS = S.SIZE_GROUPS["12b"]
```

## Selected examples

These examples were selected manually to inspect differences between SFT and DPO, including disagreements between BERTScore and judge ratings and a case where the judge rates DPO lower. They illustrate individual behaviors, not their frequency in the validation set.

```python
HIGHLIGHTS = [
    ('072_record-1734597505519_ms', 15,
     'SFT suggests giving the infant water, while DPO asks a follow-up question. The judge assigns DPO a higher safety and role-adherence rating, but SFT has the higher BERTScore. The reference also contains a feeding suggestion.'),
    ('108_record-1738229017688_ms', 23,
     'SFT asks about physical violence. DPO asks whether the behavior just described involved hitting. The judge rates DPO higher on all four rubrics.'),
    ('071_record-1734597711254_ms', 19,
     'After the parent describes reduced urination and darker urine, SFT asks about changes in behavior. DPO asks about blood in the stool or vomit and receives a higher clinical-usefulness rating.'),
    ('051_record-1734426390863_ms', 5,
     'The parent has already given the child’s age. SFT asks a question that mixes age and pacifier use; DPO asks specifically about pacifier use during sleep.'),
    ('027_TamaraTijanaUsi_ms', 19,
     'SFT repeats a question about discharge that was answered earlier and has the higher BERTScore. DPO asks a new question and receives a higher mean judge score.'),
    ('056_record_241217_ms', 21,
     'Here DPO asks about a rash that has been discussed since the start of the conversation. SFT asks about contact with someone who is ill. This example shows a decrease in the judge ratings after DPO.'),
]

for call_id, turn_id, caption in HIGHLIGHTS:
    display(S.show_item(
        items, call_id, turn_id, models=MODELS[1:],
        metrics=["bertscore_f1"], caption=caption,
    ))
```
